```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('dog.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(image)
plt.title("dog image")
plt.show()
```

# 1. Smoothing filter


```python
average_filter = cv2.blur(image, (30,30))
plt.figure(figsize = (20,10))
plt.subplot(1,2,1);plt.imshow(image);plt.title('Input Image')
plt.subplot(1,2,2);plt.imshow(average_filter);plt.title('Output Image(Average Filter)')
```

## 2. Using Weighted Averaging Filter

```python
kernel = np.array([[1,2,1],
                   [2,4,2],
                   [1,2,1]])/16
weighted_average_filter = cv2.filter2D(image, -1, kernel)
```

```python
# Display the images.
plt.figure(figsize = (20,10))
plt.subplot(1,2,1);plt.imshow(image); plt.title('Input Image')
plt.subplot(1,2,2);plt.imshow(weighted_average_filter); plt.title('Output Image(weighted_average_filter)')
```

# 3.Gaussian filter

```python
# Apply Gaussian blur.
gaussian_filter = cv2.GaussianBlur(image, (29,29), 0, 0)
```

```python
plt.figure(figsize = (20,10))
plt.subplot(1,2,1); plt.imshow(image); plt.title('Input Image')
plt.subplot(1,2,2); plt.imshow(gaussian_filter); plt.title('Output Image ( Gaussian Filter)')
```

# 4. Median Filter

```python
median_filter = cv2.medianBlur(image, 19)
```

```python
plt.figure(figsize = (18, 6))
plt.subplot(121); plt.imshow(image); plt.title('Input Image')
plt.subplot(122); plt.imshow(median_filter); plt.title('Output Image ( Median_filter)')
```

# 5. Sharpening Filters 


```python
laplacian_kernel = np.array([[0, -1, 0],
                             [-1, 5, -1],
                             [0, -1, 0]])
sharpened_laplacian_kernel = cv2.filter2D(image, -1, kernel = laplacian_kernel)
```

```python
# Display the images.

plt.figure(figsize = (18, 6))
plt.subplot(1,2,1); plt.imshow(image); plt.title('Input Image')
plt.subplot(1,2,2); plt.imshow(sharpened_laplacian_kernel); plt.title('Output Image ( Laplacian_filter)')
```

```python
gray_image = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
laplacian_operator = cv2.Laplacian(gray_image, cv2.CV_64F)
laplacian_operator = np.uint8(np.absolute(laplacian_operator))
```

```python
plt.figure(figsize = (18, 6))
plt.subplot(131); plt.imshow(image); plt.title('Input Image')
plt.subplot(132); plt.imshow(gray_image, cmap='gray'); plt.title('Gray_image')
plt.subplot(133); plt.imshow(laplacian_operator,cmap='gray'); plt.title('Output Image ( Laplacian_filter)')
```

```python
plt.figure(figsize = (18, 12))

plt.subplot(3,3,1); plt.imshow(image); plt.title('Original Image')
plt.subplot(3,3,2); plt.imshow(average_filter); plt.title('Average Filter')
plt.subplot(3,3,3); plt.imshow(weighted_average_filter); plt.title('Weighted Average Filter')

plt.subplot(3,3,4); plt.imshow(gaussian_filter); plt.title('Gaussian Filter')
plt.subplot(3,3,5); plt.imshow(median_filter); plt.title('Median Filter')
plt.subplot(3,3,6); plt.imshow(sharpened_laplacian_kernel); plt.title('Laplacian Sharpening')

plt.subplot(3,3,7); plt.imshow(gray_image, cmap='gray'); plt.title('Grayscale Image')
plt.subplot(3,3,8); plt.imshow(laplacian_operator, cmap='gray'); plt.title('Laplacian Operator')

plt.tight_layout()
plt.show()
```
